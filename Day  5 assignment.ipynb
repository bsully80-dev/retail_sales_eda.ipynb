```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
churn = pd.read_csv("customer_churn.csv")
X_cls = churn.drop(columns="Churn")
y_cls = churn["Churn"]

churn.head()
```

```python
num_cols = ["MonthlyCharges", "Tenure", "SupportCalls"]
cat_cols = ["ContractType", "PaymentMethod"]
```

```python
preprocess = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), num_cols),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), cat_cols)
])
```

```python
Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_cls, y_cls, test_size=0.20, random_state=42, stratify=y_cls
)
```

```python
logistic_pipe = Pipeline([
    ("preprocess", preprocess),
    ("model", LogisticRegression(max_iter=1000))
])
logistic_pipe.fit(Xc_train, yc_train)
```

```python
yc_pred = logistic_pipe.predict(Xc_test)
yc_prob = logistic_pipe.predict_proba(Xc_test)[:, 1]
```

```python
print("Accuracy:", accuracy_score(yc_test, yc_pred))
print("Precision:", precision_score(yc_test, yc_pred, zero_division=0))
print("Recall:", recall_score(yc_test, yc_pred, zero_division=0))
print("F1-Score:", f1_score(yc_test, yc_pred, zero_division=0))
print("ROC-AUC:", roc_auc_score(yc_test, yc_prob))
print("\nConfusion Matrix:\n", confusion_matrix(yc_test, yc_pred))
```
